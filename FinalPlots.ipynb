```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from mpl_toolkits.mplot3d import Axes3D  # needed for 3D
```

```python
# Turning phi and wind dicts to floats
def p_field(p_spec):
    return np.array(p_spec["phi"], dtype=float)

def wind_field_star(wind_spec):
    return np.array(wind_spec["u_star"], dtype=float)

# Non-dimensionalized ODE solver
def odes_star(tau, y, Gamma, Lambda, wind_spec):
    x, y_, z, vx, vy, vz = y
    v = np.array([vx, vy, vz])
    u = wind_field_star(wind_spec)

    vrel = v - u
    speed = np.linalg.norm(vrel)

    dxdt = vx
    dydt = vy
    dzdt = vz

    dvxdt = -Lambda * speed * vrel[0]
    dvydt = -Lambda * speed * vrel[1]
    dvzdt = -Gamma - Lambda * speed * vrel[2]

    return [dxdt, dydt, dzdt, dvxdt, dvydt, dvzdt]
```

```python
# Landing position
def hit_ground(tau, y):
    return y[2]

hit_ground.terminal = True
hit_ground.direction = -1
```

```python
# Initial velocity
def initial_velocity(theta, phi):
    return np.array([
        np.cos(theta)*np.cos(phi),
        np.cos(theta)*np.sin(phi),
        np.sin(theta)
    ])
```

```python
# Trajectory modelling with launch angle, launch direction, and wind speed
def simulate(theta, phi, Gamma, Lambda, wind_spec,
             z0_star=0.0, tau_max=30.0):

    y0 = np.zeros(6)
    y0[0:3] = [0.0, 0.0, z0_star]
    y0[3:6] = initial_velocity(theta, p_field(phi))

    sol = solve_ivp(
        lambda t, y: odes_star(t, y, Gamma, Lambda, wind_spec),
        (0, tau_max),
        y0,
        events=hit_ground,
        rtol=1e-4,
        atol=1e-6,
        max_step=0.05
    )

    if sol.status != 1:
        return None, None

    xL, yL = sol.y[0, -1], sol.y[1, -1]
    rL = np.hypot(xL, yL)
    psiL = np.arctan2(yL, xL)

    return sol, {"xL": xL, "yL": yL, "rL": rL, "psiL": psiL}
```

```python
Gamma  = 0.40      # stronger gravity relative to launch speed
Lambda = 0.40      # strong drag
z0_star = 0.0

theta = np.deg2rad(75) # launch angle
phi   = [
    {"phi": 0.0},
    {"phi": 2*np.pi/5},
    {"phi": 4*np.pi/5},
    {"phi": 6*np.pi/5},
    {"phi": 8*np.pi/5},
]  # different launch directions based on different spoke positions

winds = [
    {"u_star": (0.0, 0.0, 0.0)},
    {"u_star": (0.05, 0.0, 0.0)},
    {"u_star": (0.10, 0.0, 0.0)},
    {"u_star": (0.15, 0.0, 0.0)},
    {"u_star": (0.20, 0.0, 0.0)},
    {"u_star": (0.25, 0.0, 0.0)}
] # keeping and increasing wind in x direction

cmap = plt.get_cmap("Paired") # for plotting
plt.rcParams.update({'font.size': 20})
```

```python
# Basic Model (fountain top-down view)
theta_circ = np.linspace(0, 2*np.pi, 400)
plt.figure(figsize=(8,6))
plt.plot(np.cos(theta_circ), np.sin(theta_circ), label="Basin", color='darkturquoise')

for j, p in enumerate(phi):
    _, land = simulate(np.deg2rad(69), p, Gamma, Lambda, {"u_star": (0.0, 0.0, 0.0)})
    if land:
        plt.plot([0, land["xL"]], [0, land["yL"]], marker="o", label=f"spoke {j+1} (phi={p['phi']:.2f} rads)", color=cmap(2*j + 1))

plt.gca().set_aspect("equal")
plt.xlabel(r"$x^*$", fontsize=20)
plt.ylabel(r"$y^*$", fontsize=20)
#plt.title("Five-Spoke Symmetry of Nozzle Configuration (Top-Down View)", fontsize=20)
plt.legend(loc="upper left",
    bbox_to_anchor=(1.02, 1.0),
    borderaxespad=0.0, fontsize=20)

plt.savefig("fig2.pdf", bbox_inches='tight')
plt.show()
```

```python
# Jet trajectories under increasing wind, viewed along x-z plane
plt.figure(figsize=(8,6))
for i, p in enumerate(phi):
    for j, w in enumerate(winds):
        sol, _ = simulate(theta, p, Gamma, Lambda, w)
        if sol is not None:
            if i == 0:
                plt.plot(sol.y[0], sol.y[2], color=cmap(2*j + 1),
                        label=f"u*={w['u_star'][0]:.2f}")
            else:
                plt.plot(sol.y[0], sol.y[2], color=cmap(2*j + 1))

plt.xlabel(r"$x^*$", fontsize=20)
plt.ylabel(r"$z^*$", fontsize=20)
#plt.title(r"Jet Trajectories Under Increasing Wind", fontsize=20)
plt.legend(loc="upper left",
    bbox_to_anchor=(1.02, 1.0),
    borderaxespad=0.0, fontsize=20)
plt.show()
```

```python
# Jet trajectories under increasing wind, viewed along y-z plane
plt.figure(figsize=(8,6))
for i, p in enumerate(phi):
    for j, w in enumerate(winds):
        sol, _ = simulate(theta, p, Gamma, Lambda, w)
        if sol is not None:
            if i == 0:
                plt.plot(sol.y[1], sol.y[2], color=cmap(2*j + 1),
                        label=f"u*={w['u_star'][0]:.2f}")
            else:
                plt.plot(sol.y[1], sol.y[2], color=cmap(2*j + 1))

plt.xlabel(r"$y^*$", fontsize=20)
plt.ylabel(r"$z^*$", fontsize=20)
#plt.title(r"Jet Trajectories Under Increasing Wind, $\theta=0.75 \degree, \Gamma=0.4, \Lambda=0.4$", fontsize=20)
plt.legend(loc="upper left",
    bbox_to_anchor=(1.02, 1.0),
    borderaxespad=0.0, fontsize=20)
plt.show()
```

```python
# Jet trajectories under increasing wind, viewed in 3D space
fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection="3d")
ax.plot(np.cos(theta_circ), np.sin(theta_circ), label="Basin boundary", color='darkturquoise')

for i, p in enumerate(phi):
    for j, w in enumerate(winds):
        sol, _ = simulate(theta, p, Gamma, Lambda, w)
        if sol is not None:
            if i == 0:
                ax.plot(
                    sol.y[0],  # x*
                    sol.y[1],  # y*
                    sol.y[2],  # z*
                    color=cmap(2*j + 1),
                    label=f"u*={w['u_star'][0]:.2f}"
                )
            else:
                ax.plot(
                    sol.y[0],  # x*
                    sol.y[1],  # y*
                    sol.y[2],  # z*
                    color=cmap(2*j + 1),
                )

ax.set_xlabel(r"$x^*$", fontsize=20, labelpad = 16)
ax.set_ylabel(r"$y^*$", fontsize=20, labelpad = 16)
ax.set_zlabel(r"$z^*$", fontsize=20, labelpad = 16)
#ax.set_title(r"3D Fountain Trajectories Under Increasing Wind, $\theta=0.75 \degree, \Gamma=0.4, \Lambda=0.4$", fontsize=20)
ax.legend(loc="upper left",
    bbox_to_anchor=(1.1, 1.0),
    borderaxespad=0.0, fontsize=20)

plt.savefig("fig7-1.pdf", bbox_inches='tight')
plt.show()
```

```python
# Jet trajectory landing positions under increasing wind for each spoke
theta_circ = np.linspace(0, 2*np.pi, 400)
plt.figure(figsize=(8,6))
plt.plot(np.cos(theta_circ), np.sin(theta_circ), label="Basin boundary", color='darkturquoise')
for i, p in enumerate(phi):
    for j, w in enumerate(winds):
        xs, ys = [], []
        _, land = simulate(theta, p, Gamma, Lambda, w)
        if land:
            xs.append(land["xL"])
            ys.append(land["yL"])

            if i == 0:
                plt.scatter(xs, ys,
                    color=cmap(2*j + 1),
                    label=f"u*={w['u_star'][0]:.2f}")
            else:
                plt.scatter(xs, ys,
                    color=cmap(2*j + 1))

plt.gca().set_aspect("equal")
plt.xlabel(r"$x^*$", fontsize=20)
plt.ylabel(r"$y^*$", fontsize=20)
#plt.title(r"Landing Locations Under Wind Disturbances, $\theta=0.75 \degree, \Gamma=0.4, \Lambda=0.4$", fontsize=20)
plt.legend(loc="upper left",
    bbox_to_anchor=(1.02, 1.0),
    borderaxespad=0.0, fontsize=20)
plt.show()
```

```python
# Jet trajectory landing positions under increasing wind for each spoke
plt.figure(figsize=(8,6))
plt.plot(np.cos(theta_circ), np.sin(theta_circ), label="Basin Boundary", color='darkturquoise')

for i, w in enumerate(winds):
    for j, p in enumerate(phi):
        _, land = simulate(theta, p, Gamma, Lambda, w)
        if j == 0:
            if land:
                plt.plot([0, land["xL"]], [0, land["yL"]], marker="o", label=f"u*={w['u_star'][0]:.2f}", color =cmap(2*i + 1))
        else:
            if land:
                plt.plot([0, land["xL"]], [0, land["yL"]], marker="o", color =cmap(2*i + 1))

plt.gca().set_aspect("equal")
plt.xlabel(r"$x^*$", fontsize=20)
plt.ylabel(r"$y^*$", fontsize=20)
#plt.title(r"Landing Locations Under Various Wind Conditions, $\theta=0.75 \degree, \Gamma=0.4, \Lambda=0.4$", fontsize=20)
plt.legend(loc="upper left",
    bbox_to_anchor=(1.02, 1.0),
    borderaxespad=0.0, fontsize=20)

plt.savefig("fig7-2.pdf", bbox_inches='tight')
plt.show()
```

```python
# Landing radius vs launch angles
wind_fixed = {"u_star": (0.1, 0.0, 0.0)}
angles = np.deg2rad(np.linspace(0, 90, 40))

rvals = np.full((len(phi), len(angles)), np.nan)

for j, p in enumerate(phi):
    for i, th in enumerate(angles):
        _, land = simulate(th, p, Gamma, Lambda, wind_fixed)
        if land is not None:
            rvals[j, i] = land["rL"]

plt.figure(figsize=(8,6))
for j, p in enumerate(phi):
    if j == 3:
        break
    plt.plot(np.rad2deg(angles), rvals[j], label=f"spoke {j+1}", color=cmap(2*j + 1))

plt.axhline(1.0, linestyle="--", color='darkturquoise')
plt.xlabel(r"Launch Angle $\theta$ (deg)", fontsize=20)
plt.ylabel(r"Landing Radius $r_L^*$", fontsize=20)
#plt.title(r"Landing Radius vs Launch Angle (One Curve per Spoke), $\Gamma=0.4, \Lambda=0.4, u^*_x=1$", fontsize=20)
plt.legend(fontsize=20)

plt.savefig("fig3-1.pdf", bbox_inches='tight')
plt.show()
```

```python
# Landing radius vs launch angles (contd.)
plt.figure(figsize=(8,6))
for j, p in enumerate(phi):
    if j < 3:
        continue
    plt.plot(np.rad2deg(angles), rvals[j], label=f"spoke {j+1}", color=cmap(2*j + 1))

plt.axhline(1.0, linestyle="--", label="Basin boundary", color='darkturquoise')
plt.xlabel(r"Launch Angle $\theta$ (deg)", fontsize=20)
plt.ylabel(r"Landing Radius $r_L^*$", fontsize=20)
#plt.title(r"Landing Radius vs Launch Angle (One Curve per Spoke), $\Gamma=0.4, \Lambda=0.4, u^*_x=1$", fontsize=20)
plt.legend(fontsize=20)

plt.savefig("fig3-2.pdf", bbox_inches='tight')
plt.show()
```

```python
# Jet trajectories under fixed wind and increasing drag, viewed in 3D space
Lambdas = np.linspace(0.0, 2.5, 6)

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection="3d")
ax.plot(np.cos(theta_circ), np.sin(theta_circ), label="Basin boundary", color='darkturquoise')

for i, Lam in enumerate(Lambdas):
    for j, p in enumerate(phi):
        sol, _ = simulate(theta, p, Gamma, Lam, wind_fixed)
        if sol is not None:
            if j == 0:
                ax.plot(
                    sol.y[0],  # x*
                    sol.y[1],  # y*
                    sol.y[2],  # z*
                    color=cmap(2*i + 1),
                    label=f"lambda = {Lam}"
                )
            else:
                ax.plot(
                    sol.y[0],  # x*
                    sol.y[1],  # y*
                    sol.y[2],  # z*
                    color=cmap(2*i + 1),
                )

ax.set_xlabel(r"$x^*$", fontsize=20, labelpad=16)
ax.set_ylabel(r"$y^*$", fontsize=20, labelpad=16)
ax.set_zlabel(r"$z^*$", fontsize=20, labelpad=16)
#ax.set_title(r"3D Fountain Jet Trajectories Under Different Drag Strengths, $\theta = 75 \degree, \Gamma=0.4, u^*_y=2.5, u^*_z=6$", fontsize=20)
ax.legend(loc="upper left",
    bbox_to_anchor=(1.1, 1.0),
    borderaxespad=0.0, fontsize=20)

plt.savefig("fig5.pdf", bbox_inches='tight')
plt.show()
```

```python
# Basin radius vs drag
Lambdas = np.linspace(0.0, 20, 50)
rvals = np.full((len(phi), len(Lambdas)), np.nan)

for i, Lam in enumerate(Lambdas):
    for j, p in enumerate(phi):
        _, land = simulate(theta, p, Gamma, Lam, wind_fixed)
        if land is not None:
            rvals[j, i] = land["rL"]

plt.figure(figsize=(8,6))
for j, p in enumerate(phi):
    if j == 3:
        break
    plt.plot(Lambdas, rvals[j], label=f"spoke {j+1}", color=cmap(2*j + 1))
plt.axhline(1.0, linestyle="--", color='darkturquoise')
plt.xlabel(r"Drag Strength $\Lambda$", fontsize=20)
plt.ylabel(r"Landing Radius $r_L^*$", fontsize=20)
#plt.title(r"Effect of Drag on Basin Confinement, $\theta = 75 \degree, \Gamma=0.4, u^*_y=20, u^*_z=50$", fontsize=20)
plt.legend(fontsize=20)

plt.savefig("fig4-1.pdf", bbox_inches='tight')
plt.show()
```

```python
# Basin radius vs drag (contd.)
plt.figure(figsize=(8,6))
for j, p in enumerate(phi):
    if j < 3:
        continue
    plt.plot(Lambdas, rvals[j], label=f"spoke {j+1}", color=cmap(2*j + 1))
plt.axhline(1.0, linestyle="--", label="Nominal basin radius", color='darkturquoise')
plt.xlabel(r"Drag strength $\Lambda$", fontsize=20)
plt.ylabel(r"Landing radius $r_L^*$", fontsize=20)
#plt.title(r"Effect of Drag on Basin Confinement, $\theta = 75 \degree, \Gamma=0.4, u^*_y=20, u^*_z=50$", fontsize=20)
plt.legend(fontsize=20)

plt.savefig("fig4-2.pdf", bbox_inches='tight')
plt.show()
```
